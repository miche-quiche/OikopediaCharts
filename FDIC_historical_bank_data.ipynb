```python
import pandas as pd
import matplotlib.pyplot as plt
from utils.price_adjust import real_price
plt.style.use('https://github.com/dhaitz/matplotlib-stylesheets/raw/master/pacoty.mplstyle')

# Charge CSV
df = pd.read_csv("Data/FDICHistoricalDataCommercialBanksFinancial.csv")

# Tri par année au cas où
df = df.sort_values("YEAR")

# en dollar constant
df["ASSET_real"] = real_price(df["ASSET"], df["YEAR"], base=2024)/1000000

# Plot de l'actif (ASSET) au cours du temps
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(df["YEAR"], df["ASSET_real"])

ax.set_xmargin(0)
ax.set_ylim(bottom=0)
ax.set_ylabel("Milliards de dollars (2024)")
ax.set_title("Actif des banques au cours du temps")

fig.text(0.01, -0.02, "Source: FDIC Historical Bank Data", ha="left", fontsize=9)
plt.tight_layout()
plt.show()
```
